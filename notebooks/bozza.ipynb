```python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")


client = OpenAI(api_key=api_key)
```

### Variables

```python
location = "San Francisco"
list_of_vegetables_and_berries=["carrots", "raspberries", "eggplants", "habanero peppers"]
number_of_people = 4
```

### Agent Class

```python
class Agent:
    """
    A generic agent that interacts with an LLM.
    Can be customized for different domains and personalities.
    """
    def __init__(self, name="GenericAgent", model="gpt-4o-mini", role="helpful assistant"):
        self.name = name
        self.model = model
        self.role = role

    def ask(self, prompt: str, client=None):
        if client is None:
            return "Error: no client provided, use ask()."
        
        try:
            response = client.chat.completions.create(
                model=self.model,
                messages=[
                    {"role": "system", "content": f"You are {self.name}, {self.role}."},
                    {"role": "user", "content": prompt}
                ]
            )
            return response.choices[0].message.content
        except Exception as e:
            return f"Errore: {e}"
```

### Plant Selector Agent

```python

gardening_agent = Agent(
    name="PlantSelector",
    model="gpt-4o-mini",
    role=" a helpful gardening assistant specialized in synergistic gardening."
)

plant_list = gardening_agent.ask((
    
    "I live in {location}, and I like {list_of_vegetables_and_berries}. "
    "I want to make a synergistic garden. "
    "Add to my preference the most useful and productive garden plants that thrive in my climate zone"
    "Suggest a list of plants that do well in my area and that have nice synergies."
    "Avoid any tips or advice, generate only a list of plants. Avoid any explanations or introduction."
).format(
    location=location,
    list_of_vegetables_and_berries=list_of_vegetables_and_berries
), client=client )

```

```python
print(plant_list)
```

### Synergy Agent

```python
synergy_agent = Agent(
    name="SynergyAgent",
    model="gpt-4o-mini",
    role=" a helpful gardening assistant specialized in synergistic gardening."
)
plants_and_synergy =synergy_agent.ask((
    "Based on this list of plant: {plant_list}, provide any additional synergetic plant that thrives in this location: {location}"
    "Return the provided list with the addition of the new plants."
    "Avoid any tips or advice, generate only a list of plants. Avoid any explanations or introduction."
).format(
    plant_list=plant_list,
    location=location
), client=client)


```

```python

print(plants_and_synergy)
```

### Garden Agent

```python
garden_agent = Agent(
    name="GardenAgent",
    model="gpt-4o-mini",
    role=" a helpful gardening assistant specialized in synergistic gardening."
)
gardening_instruction =garden_agent.ask((
    "Based on this list of plant: {plant_list}, provide instructions to set the garden to take full advantage of the synergies between plants"
    "Provide precise instructions on which plants should be planted close to each other and on the synergies to take advantage of."
    "Give the right number of plants for each species to feed a family of {number_of_people}"
).format(
    plant_list=plant_list,
    number_of_people=number_of_people,
), client=client)
```

```python
print(gardening_instruction)
```
