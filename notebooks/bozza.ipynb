```python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()
api_key = os.getenv("OPENAI_API_KEY")

client = OpenAI(api_key=api_key)
```

```python

class PlantSelector:
    """
    A simple agent interacting with an LLM
    """
    def __init__(self, name="PlantSelector", model="gpt-4o-mini"):
        self.name = name
        self.model = model

    def ask(self, prompt: str):
        try:
            response = client.chat.completions.create(
                model=self.model,
                messages=[
                    {"role": "system", "content": f"You are {self.name}, a helpful gardening assistant specialized in synergistic gardening."},
                    {"role": "user", "content": prompt}
                ]
            )
            return response.choices[0].message.content
        except Exception as e:
            return f"Errore: {e}"
```

```python
location = "San Francisco"
list_of_vegetables_and_berries=["carrots", "raspberries", "eggplants", "habanero peppers"]
```

```python
agent = PlantSelector("GardenBot")
prompt = (
    "I live in {location}, and I like {list_of_vegetables_and_berries}. "
    "I want to make a synergistic garden. "
    "Add to my preference the most useful and productive garden plants that thrive in my climate zone"
    "Suggest a list of plants that do well in my area and that have nice synergies."
    "Avoid any tips or advice, generate only a list of plants. Avoid any explanations or introduction."
).format(
    location=location,
    list_of_vegetables_and_berries=list_of_vegetables_and_berries
)

plant_list = agent.ask(prompt)
print(plant_list)
```
